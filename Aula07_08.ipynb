```python
img = cv2.imread('Radiohead_-_In_Rainbows.jpg')
B, G, R = cv2.split(img)
cv2_imshow(B)

# Imagens rgb possuem 3 camadas. 1 camada: Bitlab Vermelho, 2 Camada: Bitlab verde, 3 Camada: Bitlab Azul
# Se for extraida apenas uma camada ele entende como uma imagem preto e branco
# Ele vai ler o valor dos pixel como intencidade do branco da imagem
```

```python
#Subplot azul
plt.subplot(1, 3, 1)
plt.imshow(B, cmap = 'gray')
plt.axis('off')
plt.title('Azul(B)')

# Subplot Verde
plt.subplot(1, 3, 2)
plt.imshow(G, cmap = 'gray')
plt.axis('off')
plt.title('Verde(G)')


# Subplot Vermelho
plt.subplot(1, 3, 3)
plt.imshow(R, cmap = 'gray')
plt.axis('off')
plt.title('Vermelho(R)')


```
