```python
# 1 - ler imagem
# 2 - converte pra cinza
# 3 - percorre a matriz da imagem e p/ cada pixel:
## identificar a faixa de intensidade
## incrementar a posição correspondente no valor

def calc_hist(img):
  img_vetor = img.ravel() # desmembra a matriz em um vetor
  hist_vetor = np.zeros(256)
  for px in img_vetor:
    hist_vetor[px] += 1

  return hist_vetor

```
